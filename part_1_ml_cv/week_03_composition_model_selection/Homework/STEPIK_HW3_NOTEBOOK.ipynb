## Применение линейных моделей (3 балла)

1) Обработайте данные для того, чтобы к ним можно было применить LogisticRegression. Т.е. отнормируйте числовые признаки, а категориальные закодируйте с помощью one-hot-encoding'а.

2) С помощью кроссвалидации или разделения на train/valid выборку протестируйте разные значения гиперпараметра C и выберите лучший (можно тестировать С=100, 10, 1, 0.1, 0.01, 0.001) по метрике ROC-AUC.

Если вы разделяете на train/valid, то используйте LogisticRegressionCV. Он сам при вызове .fit() подберет параметр С. (не забудьте передать scroing='roc_auc', чтобы при кроссвалидации сравнивались значения этой метрики, и refit=True, чтобы при потом модель обучилась на всем датасете с лучшим параметром C).


(более сложный вариант) Если вы будете использовать кроссвалидацию, то преобразования данных и LogisticRegression нужно соединить в один Pipeline с помощью make_pipeline, как это делалось во втором семинаре. Потом pipeline надо передать в GridSearchCV. Для one-hot-encoding'a можно испльзовать комбинацию LabelEncoder + OneHotEncoder (сначала превращаем строчки в числа, а потом числа првращаем в one-hot вектора.)

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score
from sklearn.model_selection import GridSearchCV, StratifiedKFold, train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
```

```python
X = data[feature_cols].copy()
y = data[target_col].copy()

preprocessor = ColumnTransformer([
    ("numeric", Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
    ]), num_cols),
    ("categorical", OneHotEncoder(handle_unknown="ignore"), cat_cols),
])

logistic_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(max_iter=2000, random_state=42)),
])

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
logistic_search = GridSearchCV(
    logistic_pipeline,
    param_grid={"model__C": [0.01, 0.1, 1, 10, 100]},
    scoring="roc_auc",
    cv=cv,
    n_jobs=1,
    refit=True,
)
logistic_search.fit(X, y)

logistic_results = pd.DataFrame(logistic_search.cv_results_)[
    ["param_model__C", "mean_test_score", "std_test_score"]
].sort_values("mean_test_score", ascending=False)
display(logistic_results)
print("Лучший C:", logistic_search.best_params_["model__C"])
print(f"ROC-AUC логистической регрессии: {logistic_search.best_score_:.4f}")
```

Логистическая регрессия оценивалась стратифицированной 5-fold кросс-валидацией. Лучшее качество получилось при значении `C`, напечатанном выше; средний ROC-AUC находится примерно на уровне **0.844**. Все преобразования обучались внутри каждого фолда, поэтому утечки из валидации нет.


## Применение градиентного бустинга (2 балла)

Если вы хотите получить баллы за точный ответ, то стоит попробовать градиентный бустинг. Часто градиентный бустинг с параметрами по умолчанию даст вам 80% результата за 0% усилий.

Мы будем использовать catboost, поэтому нам не надо кодировать категориальные признаки. Catboost сделает это сам (в .fit() надо передать cat_features=cat_cols). А численные признаки нормировать для моделей, основанных на деревьях не нужно.

1) Разделите выборку на train/valid. Протестируйте catboost cо стандартными параметрами.

2) Протестируйте разные занчения параметров количества деревьев и learning_rate'а и выберите лучшую по метрике ROC-AUC комбинацию.

(Дополнительно) Есть некоторые сложности с тем, чтобы использовать CatBoostClassifier вместе с GridSearchCV, поэтому мы не просим использовать кроссвалидацию. Но можете попробовать)

```python
from catboost import CatBoostClassifier

X_cat = data[feature_cols].copy()
y_cat = data[target_col].copy()
X_train, X_valid, y_train, y_valid = train_test_split(
    X_cat, y_cat, test_size=0.25, random_state=42, stratify=y_cat
)

median_total_spent = X_train["TotalSpent"].median()
X_train["TotalSpent"] = X_train["TotalSpent"].fillna(median_total_spent)
X_valid["TotalSpent"] = X_valid["TotalSpent"].fillna(median_total_spent)
for column in cat_cols:
    X_train[column] = X_train[column].astype(str)
    X_valid[column] = X_valid[column].astype(str)

catboost_configs = [
    {"depth": 4, "learning_rate": 0.03, "iterations": 800, "l2_leaf_reg": 5},
    {"depth": 5, "learning_rate": 0.03, "iterations": 800, "l2_leaf_reg": 5},
    {"depth": 6, "learning_rate": 0.03, "iterations": 800, "l2_leaf_reg": 5},
    {"depth": 5, "learning_rate": 0.05, "iterations": 600, "l2_leaf_reg": 7},
]

catboost_results = []
best_catboost = None
best_catboost_auc = -np.inf
for config in catboost_configs:
    model = CatBoostClassifier(
        **config,
        loss_function="Logloss",
        eval_metric="AUC",
        random_seed=42,
        verbose=False,
        allow_writing_files=False,
    )
    model.fit(
        X_train,
        y_train,
        cat_features=cat_cols,
        eval_set=(X_valid, y_valid),
        early_stopping_rounds=100,
        verbose=False,
    )
    predictions = model.predict_proba(X_valid)[:, 1]
    auc = roc_auc_score(y_valid, predictions)
    catboost_results.append({**config, "best_iteration": model.get_best_iteration(), "valid_auc": auc})
    if auc > best_catboost_auc:
        best_catboost_auc = auc
        best_catboost = model

display(pd.DataFrame(catboost_results).sort_values("valid_auc", ascending=False))
print(f"Лучший ROC-AUC CatBoost на отложенной выборке: {best_catboost_auc:.4f}")
```

Лучший результат на отложенной выборке дал CatBoost небольшой глубины с низким `learning_rate` и ранней остановкой: ROC-AUC около **0.848**. # Предсказания

```python
# Обучаем финальные модели на всех доступных данных.
best_logistic = logistic_search.best_estimator_

X_cat_full = data[feature_cols].copy()
full_total_spent_median = X_cat_full["TotalSpent"].median()
X_cat_full["TotalSpent"] = X_cat_full["TotalSpent"].fillna(full_total_spent_median)
for column in cat_cols:
    X_cat_full[column] = X_cat_full[column].astype(str)

final_catboost_models = []
for seed in [42, 123, 777]:
    model = CatBoostClassifier(
        iterations=330,
        depth=4,
        learning_rate=0.03,
        l2_leaf_reg=5,
        loss_function="Logloss",
        eval_metric="AUC",
        random_seed=seed,
        verbose=False,
        allow_writing_files=False,
    )
    model.fit(X_cat_full, y, cat_features=cat_cols, verbose=False)
    final_catboost_models.append(model)
```

```python
X_test = pd.read_csv('./test.csv')
submission = pd.read_csv('./submission.csv')

X_test["TotalSpent"] = pd.to_numeric(X_test["TotalSpent"], errors="coerce")
logistic_test_predictions = best_logistic.predict_proba(X_test[feature_cols])[:, 1]

X_test_cat = X_test[feature_cols].copy()
X_test_cat["TotalSpent"] = X_test_cat["TotalSpent"].fillna(full_total_spent_median)
for column in cat_cols:
    X_test_cat[column] = X_test_cat[column].astype(str)

catboost_test_predictions = np.mean([
    model.predict_proba(X_test_cat)[:, 1]
    for model in final_catboost_models
], axis=0)

submission["Churn"] = 0.6 * catboost_test_predictions + 0.4 * logistic_test_predictions
submission.to_csv('./my_submission.csv', index=False)

display(submission.head())
print("Размер посылки:", submission.shape)
print("Диапазон вероятностей:", submission["Churn"].min(), submission["Churn"].max())
```
